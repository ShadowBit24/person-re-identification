```python
import zipfile

zip_path = "Spatial-Temporal-Re-identification-1.zip"

ext_dir = "./"

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(ext_dir)

print(f"Zip file '{zip_path}' extracted to '{ext_dir}'.")
```

```python
import shutil
import os

# Specify the source and destination directories
source_dir = 'Spatial-Temporal-Re-identification-1'
destination_dir = './'

# List all files and folders in the source directory
contents = os.listdir(source_dir)

# Move each file and folder to the destination directory
for item in contents:
    source_item = os.path.join(source_dir, item)
    destination_item = os.path.join(destination_dir, item)
    shutil.move(source_item, destination_item)

print(f"All contents from '{source_dir}' moved to '{destination_dir}'.")
```
